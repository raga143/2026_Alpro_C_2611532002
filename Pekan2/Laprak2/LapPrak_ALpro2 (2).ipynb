## 2.1 Dasar Teori

1. Tipe Data Boolean
Boolean (`bool`) adalah tipe data paling sederhana dalam logika pemrograman karena hanya mengenal dua kemungkinan nilai, yakni `True` (benar) dan `False` (salah) (Python Software Foundation, 2026).

    - Pengujian Kondisi: Nilai boolean biasanya muncul dari hasil operasi perbandingan seperti `>=`, `<=`, `==`, atau `!=`.

    - Ekspresi Logika: Beberapa nilai boolean dapat digabungkan lewat operator logika `and`, `or`, dan `not` untuk mengambil keputusan dalam alur program.

    - Pengendalian Alur (Control Flow): Boolean menjadi syarat utama pada struktur percabangan (`if-else`) untuk menentukan blok kode yang akan dijalankan.

2. Tipe Data Input dan Konversi Tipe Data (Type Casting)
Input adalah cara program berinteraksi dengan pengguna untuk menerima data secara langsung ke dalam memori (Microsoft, 2026).

    - Fungsi `input()`: Dipakai untuk menangkap masukan teks dari pengguna lewat terminal/console.

    - Sifat Luaran Teks (String Default): Seluruh hasil dari `input()` otomatis tersimpan sebagai tipe `str`, sekalipun pengguna mengetikkan angka.

    - Konversi Tipe Data (Type Casting): Karena hasil `input()` selalu berbentuk string, dibutuhkan konversi eksplisit lewat fungsi seperti `int()` atau `float()` supaya nilai tersebut bisa diolah dalam operasi hitung.

3. Tipe Data Konstanta (Constant)
Python secara bawaan belum menyediakan mekanisme untuk mengunci nilai suatu variabel secara mutlak layaknya konstanta pada bahasa pemrograman lain.

    - Konvensi Penulisan: Kebiasaan yang lazim dipakai adalah menuliskan nama variabel bernilai tetap dengan huruf kapital semua (contoh: `PI`).

    - Penggunaan `typing.Final`: Modul `Final` dari pustaka `typing` dapat dimanfaatkan untuk menandai secara eksplisit bahwa sebuah variabel bersifat konstan dan tidak boleh diubah selama program berjalan (Python Software Foundation, 2026).

    - Keamanan Kode: Menjaga nilai-nilai penting, seperti konstanta matematis atau pengaturan sistem, agar tidak tertimpa secara tidak sengaja oleh baris kode lainnya.

4. Tipe Data Numerik
Tipe data numerik dipakai untuk menampung nilai berupa angka. Python mengenal tiga kategori utama tipe numerik:

    - `Integer` (`int`): Menyimpan bilangan bulat, baik positif maupun negatif, tanpa titik desimal (contoh: `100`, `-5`).

    - `Floating-Point` (`float`): Menyimpan bilangan pecahan/desimal yang ditulis dengan tanda titik (.) (contoh: `10.2345`).

    - `Complex` (`complex`): Menyimpan bilangan kompleks, di mana bagian imajinernya ditandai dengan huruf `j` di akhir (contoh: `100 + 3j`).

5. Tipe Data String
`String` (`str`) merupakan rangkaian karakter yang tersusun dari huruf, angka, simbol, ataupun spasi (Python Software Foundation, 2026).

    - Petik Tunggal (`' '`) / Petik Ganda (`" "`): Dipakai untuk menulis kalimat atau teks dalam satu baris.

    - Petik Tiga (`""" """`): Dipakai untuk menulis teks yang terdiri atas beberapa baris (multiline string), sekaligus mempertahankan format baris baru dan indentasi di dalamnya.

    - Identifikasi Tipe Data (`type()`): Fungsi bawaan `type()` berguna untuk mengecek tipe data dari suatu variabel yang sedang digunakan.
## 2.2 Kode Program

### 2.2.1 Program Numeric_2611532002
</div>


```python
#create a variable with integer value
a_2002=100
print("The type of variabel having value", a_2002, "is",type(a_2002))

#create a variabel with float value.
b_2002=10.2345
print("The type of variabel having value", b_2002, "is",type(b_2002))

#create a variabel eith complex value.
c_2002=100+3j
print("The type of variabel having value", c_2002, "is",type(c_2002))
```

Penjelasan:

3. Membuat variabel `b_2002 = 10.2345` yang bertipe Float.

4. Menampilkan nilai `b_2002` beserta tipe datanya melalui fungsi `type(b_2002)`.

5. Membuat variabel `c_2002 = 100+3j` yang bertipe Complex.

### 2.2.4 Program Input_2611532002

```python
# Buat program untuk menghitung luas persegi panjang
# Program ini menggunakan fungsi input() ubtuk meminta pengguna
# Memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

panjang_2002 = input('Masukkan nilai panjang: ')
lebar_2002 = input('Masukkan nilai lebar: ')
luas_2002 = int(panjang_2002) * int(lebar_2002)

print("Luas =", luas_2002)
```

3. Mengonversi tipe data (type casting) `panjang_2002` dan `lebar_2002` menjadi int, mengalikan keduanya, kemudian hasilnya disimpan ke variabel `luas_2002`.

4. Menampilkan (`"Luas ="`, `luas_2002`).

### 2.2.5 Program Konstanta_2611532002
</div>


```python
# Program ini menggunakan konstanta untuk menghitun luas lingkaran

from typing import Final
PI: Final = 3.14
print ("pi: %f" % (PI))
jari_2002 = float(input('masukkan nilai jari jari: '))
luas_2002 = PI * jari_2002 * jari_2002
print ("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2002, luas_2002))
```

1. Mengimpor `Final` dari modul `typing`.

2. Membuat variabel konstanta `PI: Final = 3.14` yang bertipe Float.

3. Menampilkan nilai `PI` dalam format bilangan desimal/float (`"pi: %f" % (PI)`).

4. Menerima masukan nilai jari-jari dari pengguna lewat `input()`, mengonversinya ke Float, lalu menyimpannya ke variabel `jari_2002`.
