## 2.1 Dasar Teori

Praktikum kali ini membahas pengenalan dasar terhadap bahasa pemrograman Python, penggunaan Visual Studio Code sebagai code editor, serta GitHub sebagai sarana penyimpanan dan pengelolaan kode program.

1. Python

Python adalah bahasa pemrograman tingkat tinggi dengan sintaks yang sederhana sehingga relatif mudah dipelajari oleh pemula. Bahasa ini banyak digunakan untuk berbagai keperluan, seperti pembuatan aplikasi, pengolahan data, hingga pengembangan perangkat lunak secara umum.

2. Visual Studio Code

Visual Studio Code (VS Code) merupakan editor kode yang digunakan untuk menulis, mengedit, sekaligus menjalankan program. Editor ini mendukung banyak bahasa pemrograman termasuk Python, dan dilengkapi fitur seperti terminal bawaan, debugging, serta integrasi langsung dengan Git.

Git adalah sistem kontrol versi yang berfungsi mencatat setiap perubahan yang terjadi pada kode program. Dengan Git, seorang programmer dapat mengelola berbagai versi program serta menelusuri riwayat perubahan yang pernah dilakukan.

GitHub merupakan platform daring untuk menyimpan dan mengelola repository berbasis Git. Melalui GitHub, programmer dapat melakukan pencadangan kode, mengakses proyek dari perangkat berbeda, serta berkolaborasi dengan programmer lain dalam satu proyek yang sama.

5. Repository

Repository adalah wadah penyimpanan proyek yang memuat kode program beserta berkas-berkas pendukung lainnya. Repository dapat disimpan secara lokal di komputer maupun secara daring melalui GitHub.

6. Commit dan Push

Commit merupakan proses menyimpan perubahan yang telah dibuat ke dalam repository lokal, sedangkan push digunakan untuk mengirimkan perubahan tersebut dari repository lokal menuju repository di GitHub.

### 2.2.2 Membuat Repository Github

Setelah program dibuat dan diuji pada VS Code, proyek dihubungkan ke platform GitHub. Repositori ini berfungsi untuk menyimpan, mengelola, dan mengamankan berkas kode program secara daring (online).

Tahapan yang dilakukan yaitu:

1.Membuat repositori baru pada halaman GitHub.

2.Membuka folder proyek menggunakan VS Code.

3.Membuka terminal terintegrasi pada VS Code.

4.Menginisialisasi Git pada folder proyek lokal.

5.Menambahkan berkas ke dalam staging area.

6.Membuat commit untuk menyimpan riwayat perubahan.

7.Menghubungkan repositori lokal dengan repositori di GitHub.

8.Melakukan push untuk mengunggah kode ke GitHub.

Contoh perintah (command) yang digunakan pada terminal:

1. git init: Digunakan untuk membuat dan menginisialisasi repositori Git baru pada folder proyek lokal.
2. git add .: Digunakan untuk menambahkan seluruh berkas dan perubahan pada proyek ke dalam staging area.
3. git commit: Digunakan untuk menyimpan riwayat perubahan dari staging area disertai pesan deskriptif.
4. git branch -M main: Digunakan untuk mengubah atau mengonfirmasi nama cabang (branch) utama menjadi main.
5. git remote add origin: Digunakan untuk menghubungkan repositori lokal dengan repositori yang telah dibuat di GitHub.
6. git push: Digunakan untuk mengunggah berkas dan riwayat commit dari komputer lokal ke reposi

```python
%%bash
# Inisialisasi repository baru
git init

# Cek status perubahan file
git status

# Menambahkan seluruh file ke staging area
git add .

# Menyimpan perubahan dengan pesan commit
git commit -m "Initial commit"

# Mengubah nama branch utama menjadi main
git branch -M main

# Menghubungkan repository lokal dengan repository di GitHub
git remote add origin https://github.com/username/nama-repository.git

# Mengunggah kode ke GitHub
git push -u origin main

# Mengunduh perubahan terbaru dari GitHub ke lokal
git pull origin main

# Meng-clone repository dari GitHub ke komputer lokal
git clone https://github.com/username/nama-repository.git

# Melihat riwayat commit
git log
```
